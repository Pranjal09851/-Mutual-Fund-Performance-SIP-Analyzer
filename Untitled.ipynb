```python
import requests, csv, time

FUNDS = [
    "UTI Nifty 50 Index Fund Direct Growth",
    "HDFC Index Fund Nifty 50 Plan Direct Growth",
    "Parag Parikh Flexi Cap Fund Direct Growth",
    "Axis Large Cap Fund Direct Growth",
    "Mirae Asset Large Cap Fund Direct Growth",
    "ICICI Prudential Bluechip Fund Direct Growth",
    "Kotak Emerging Equity Fund Direct Growth",
    "HDFC Mid Cap Fund Direct Growth",
    "Motilal Oswal Midcap Fund Direct Growth",
    "SBI Small Cap Fund Direct Growth",
    "Nippon India Small Cap Fund Direct Growth",
    "Axis Small Cap Fund Direct Growth",
    "Quant Small Cap Fund Direct Growth",
    "SBI Contra Fund Direct Growth",
    "Axis ELSS Tax Saver Fund Direct Growth",
]

funds_rows, nav_rows, seen = [], [], set()

for name in FUNDS:
    try:
        res = requests.get("https://api.mfapi.in/mf/search", params={"q": name}, timeout=30).json()
        match = next((r for r in res
                      if "direct" in r["schemeName"].lower()
                      and "growth" in r["schemeName"].lower()
                      and "idcw" not in r["schemeName"].lower()
                      and "dividend" not in r["schemeName"].lower()), None)
        if not match:
            print("NOT FOUND:", name); continue
        code = int(match["schemeCode"])
        if code in seen: continue
        seen.add(code)

        d = requests.get(f"https://api.mfapi.in/mf/{code}", timeout=60).json()
        m = d["meta"]
        funds_rows.append([code, m["scheme_name"], m["fund_house"],
                           m["scheme_category"], m["scheme_type"]])
        for row in d["data"]:
            day, month, year = row["date"].split("-")
            nav = float(row["nav"])
            if nav > 0:
                nav_rows.append([code, f"{year}-{month}-{day}", nav])
        print("OK:", m["scheme_name"], len(d["data"]), "rows")
        time.sleep(1)
    except Exception as e:
        print("ERROR:", name, e)

with open("funds.csv", "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    w.writerow(["scheme_code","scheme_name","fund_house","scheme_category","scheme_type"])
    w.writerows(funds_rows)

with open("nav.csv", "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    w.writerow(["scheme_code","nav_date","nav"])
    w.writerows(nav_rows)
```
